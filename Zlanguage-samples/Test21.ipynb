```python
from typing import Final

name: Final = "PyCharm"
values = [1, 2, 3, 4, 5]

def summarize(numbers: list[int]) -> dict[str, int]:
    return {"count": len(numbers), "total": sum(numbers)}

summary = summarize(values)
print("Hello, {}!".format(name))
print("Values:", values)
print("Summary:", summary)
```
